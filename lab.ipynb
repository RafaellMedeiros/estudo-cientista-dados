```python
gols_marcados = [2, 1, 3, 1, 0, 2]
gols_sofridos = [1, 2, 2, 1, 3, 0]
pontos = 0
posiveis = len(gols_marcados)*3

for i in range(len(gols_marcados)):
  gols_time = gols_marcados[i]
  gols_oponente = gols_sofridos[i]

  if(gols_time > gols_oponente):
    pontos += 3
  elif(gols_time < gols_oponente):
    continue
  else:
    pontos += 1

aprov = round(pontos/posiveis, 1)*100
print(f"A pontuação do time foi de {pontos} e seu aproveitamento foi de {aprov}%")
```

```python
infos = {
  'Salvador': {'km': 850, 'mult': 200},
  'Fortaleza': {'km': 800, 'mult': 400},
  'Fortaleza': {'km': 300, 'mult': 250},
  'Fortaleza': {'km': 550, 'mult': 300}
}

def cal_gasto_hotel(dias):
  return dias * 150

def cal_gasto_gasolina(cidade):
  km = infos[cidade]['km']
  litros = km/14
  return litros * 5

def cal_gastos_alimentacao(dias, cidade):
  mult = infos[cidade]['mult']
  return mult*dias


dias = 5
cidade = 'Salvador'
gastos = round(cal_gasto_hotel(dias) + cal_gasto_gasolina(cidade) + cal_gastos_alimentacao(dias, cidade), 2)
print(f"Com base nos gastos definidos, uma viagem de {dias} dias para {cidade} saindo de Recife custaria {gastos} reais")

```
